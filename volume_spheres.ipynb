```python
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

import matplotlib.pyplot as plt

# Create a 3D grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
z = np.linspace(-5, 5, 100)
X, Y, Z = np.meshgrid(x, y, z)

# Define two Gaussian spheres
def gaussian_sphere(X, Y, Z, center, sigma):
    return np.exp(-((X-center[0])**2 + (Y-center[1])**2 + (Z-center[2])**2) / (2*sigma**2))

sphere1 = gaussian_sphere(X, Y, Z, center=[-2, 0, 0], sigma=1)
sphere2 = gaussian_sphere(X, Y, Z, center=[2, 0, 0], sigma=1)

# Plot the spheres
print(sphere1.shape, sphere2.shape)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[sphere1 > 0.1], Y[sphere1 > 0.1], Z[sphere1 > 0.1], color='blue', alpha=0.5, label='Sphere 1')
ax.scatter(X[sphere2 > 0.1], Y[sphere2 > 0.1], Z[sphere2 > 0.1], color='red', alpha=0.5, label='Sphere 2')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('3D Gaussian Spheres')
ax.legend()
```
